```python
import caf.base as cb
from pathlib import Path
import pandas as pd
```

```python
out_dir = Path(r"E:\tem\inputs")
tfn_at = cb.ZoningSystem.get_zoning('tfn_at')
```

```python
mts_prod = pd.read_csv(r"I:\NTS\outputs\productions\hb\mode_time_splits\mode_time_split_production_hb_fr_reg.csv", index_col=[0,1,2,3,4])['rho'].unstack(level='tfn_at')
```

```python
mts.index.names = ['hh_type','p','m','tp']
mts.columns.name = 'tfn_at_id'
seg_input = cb.SegmentationInput(enum_segments=['hh_type','p','m','tp'], naming_order=['hh_type','p','m','tp'])
mts = cb.DVector(import_data=mts, segmentation=cb.Segmentation(seg_input), zoning_system=tfn_at)
```

```python
mts.save(out_dir / "mts_prod.dvec")
```

```python
mts_attr_fr = pd.read_csv(r"I:\NTS\outputs\attractions\hb\mode_time_splits\mode_time_split_attraction_hb_fr_reg.csv", index_col=[0,1,3,4])
```

```python
mts_attr_fr
```

```python
mts_attr_no_uni = mts_attr_fr[mts_attr_fr['uni']==0]['rho'].unstack(level='tfn_at')
mts_attr_uni = mts_attr_fr[mts_attr_fr['uni']==1]['rho'].unstack(level='tfn_at')
```

```python
seg = ['p','m','tp']
seg_in = cb.SegmentationInput(enum_segments=seg, naming_order=seg)
mts_attr_no_uni.index.names = seg
mts_attr_no_uni.columns.name = 'tfn_at_id'
mts_attr_no_uni = cb.DVector(import_data=mts_attr_no_uni, segmentation=cb.Segmentation(seg_in), zoning_system=tfn_at)
mts_attr_uni.index.names = seg
mts_attr_uni.columns.name = 'tfn_at_id'
mts_attr_uni = cb.DVector(import_data=mts_attr_uni, segmentation=cb.Segmentation(seg_in), zoning_system=tfn_at)
```

```python
mts_attr_no_uni.save(out_dir / "hb_attr_mts.dvec")
mts_attr_uni.save(out_dir / "hb_attr_mts_uni.dvec")
```
