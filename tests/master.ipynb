```python
# ## GENERATE PURE DEMAND ## #
trip_rates = cb.DVector.load(NHBProd.trip_rates_path)
trip_rates = trip_rates.translate_zoning(cb.ZoningSystem.get_zoning("normits"), check_totals=False, no_factors=True) # TODO pass model and get _zoning_system attribute

hbattr =  cb.DVector.load(NHBProd.hb_attraction_paths[2023])
segs = hbattr.segmentation.names
if "tp" in segs: 
    segs.remove("tp")
    hbattr = hbattr.aggregate(segs)
hbattr_data = hbattr.data.reset_index().rename(columns={"p": "p_hb", "m": "m_hb"})
segs = hbattr.segmentation.names
segs.remove("p")
segs.remove("m")
segs.append("p_hb")
segs.append("m_hb")
hbattr_data = hbattr_data.set_index(segs)
hbattr = cb.DVector(segmentation=cb.Segmentation(cb.SegmentationInput(enum_segments=segs,
                                                                      naming_order=segs)),
                    import_data=hbattr_data,
                    zoning_system=hbattr.zoning_system)

mts = cb.DVector.load(NHBProd.mts_path)
mts = mts.translate_zoning(cb.ZoningSystem.get_zoning("normits"), check_totals=False, no_factors=True)

prod_dict: dict[int, dict[int, cb.DVector]] = {}
for p in hbattr.data.reset_index()["p_hb"].unique():
    prod_dict[p] = {}
    hbattr_p = hbattr.filter_segment_value("p_hb", [p])
    tr_p = trip_rates.filter_segment_value("p_hb", [p])
    for m in hbattr_p.data.reset_index()["m_hb"].unique():
        hbattr_pm = hbattr_p.filter_segment_value("m_hb", [m])
        tr_pm = tr_p.filter_segment_value("m_hb", [m])
        prod_dict[p][m] = hbattr_pm * tr_pm

mts_dict: dict[int, dict[int, cb.DVector]] = {}
for p in prod_dict.keys():
    mts_dict[p] = {}
    for m in prod_dict[p].keys():
        prod_pm = prod_dict[p][m]
        segs = prod_pm.segmentation.names
        segs.remove("p_hb")
        segs.remove("m_hb")
        prod_pm = prod_pm.aggregate(segs)
        mts_dict[p][m] = mts * prod_pm
```
