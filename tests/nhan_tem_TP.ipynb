```python
def settlement(self, col_type: str = 'hhold') -> Dict:
        # ruc 2011 to area type
        set_01id = self.dct_to_specs(self.set_01id)
        out_dict = {
            "col": f"{col_type}ruc2011_b01id",
            "typ": str,
            "log": f"{col_type}ruc_2011",
            "val": {
                1: set_01id["major"],
                2: set_01id["minor"],
                3: set_01id["city"],
                4: set_01id["town"],
                5: set_01id["village"],
            },
            "out": {
                1: "Major",
                2: "Minor",
                3: "City",
                4: "Town",
                5: "Village",
            },
        }
        out_dict["val"] = fun.itm_to_key(out_dict["val"])
        return out_dict
```
