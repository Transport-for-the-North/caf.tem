8.6.	Update NoTEM Trip-Ends
The sequence of matrices that are produced from the Tour model above are then merged together to create a final set of matrices by mode, purpose, direction, time period at the aggregate TAZ level. They will then be used as a form of constraints to adjust the trip-end data from the NoTEM.
The NoTEM input hb_productions, hb_attractions, nhb_productions and nhb_attractions (at MSOA) level is aggregate to the same TAZ level and adjustments are to be calculated and applied to the input NoTEM zonal trip-ends to produce adjusted mode/time specific trip-ends data at MSOA level. 
It is noted that the NoTEM input trip-ends required for this process are hb_productions and hb_attractions data and only segmented by MSOA and purpose (i.e. mode agnostic demand total at “all week” level). The input NoTEM starting HB productions & attraction trip-ends will be further disaggregated by mode & time specific using the outputs from the adjustment process at the TAZ level described above. This is achieved by applying the following steps.
Step 1 - HB from home trip-ends
	Import NoTEM production & attraction trip-ends at MSOA level & aggregate to mode/time agnostic level.
	Disaggregate NoTEM trip-ends to mode and time specific using mode/time proportional splits that are calculated from the Tour model at the aggregate TAZ level.
	Calculate adjustment to MSOA level to the target trip-ends at the TAZ level.
	Output production & attraction trip-ends data at MSOA level.
Step 2 - NHB trip-ends
	Import HB attraction trip-ends that are produced from Step 1 above and aggregate to “all modes”, “all week” level (i.e. mode and time agnostic). This is undertaken due to the fact that a trip whilst arriving at a destination using a specific mode at a specific time does not necessarily leaving that zone in the same mode at the same period. The HB attraction trip-ends will be used for the disaggregation purpose of the NHB prod/attraction trip-ends.
	Disaggregate NoTEM trip-ends to mode and time specific using mode/time proportional splits that are calculated from Tour model at the aggregate TAZ level.
	Calculate adjustment to MSOA level to the target trip-ends at the TAZ level.
	Output NHB production & attraction trip-ends data at MSOA level.
Step 3 - HB return home trip-ends
	Import NHB attraction & HB production trip-ends that are produced from step 2 & step 1 respectively and aggregate to “all modes”, “all week” level (i.e. mode and time agnostic). This is undertaken due to the fact that a trip whilst arriving at a destination using a specific mode at a specific time does not necessarily leaving that zone in the same mode at the same period. NHB attractions will be used to derive HB return home origins (attractions), and HB from home production will be used to derive HB return home destinations (productions).
	Disaggregate NoTEM trip-ends to mode and time specific using mode/time proportional splits that are calculated from the Tour Model at the aggregate TAZ level.
	Calculate adjustment to MSOA level to the target trip-ends at the TAZ level.
	Output HB return home origin & destination trip-ends data at MSOA level. Origin & destination will then be transposed to convert to PA format for the Gravity model purpose (if needed).
